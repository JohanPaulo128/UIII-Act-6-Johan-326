```python
print("--- Simulador de IA: Evaluador de Crédito ---")
continuar = "si"

while continuar == "si":
    ingresos = int(input("Ingresa tus ingresos mensuales: "))
    edad = int(input("Ingresa tu edad: "))

    if edad < 18:
        print("Lo siento, debes ser mayor de edad.")
    elif ingresos > 5000:
        print("¡Felicidades! Calificas para la Tarjeta Platinum.")
    elif ingresos >= 2000:
        print("Calificas para la Tarjeta Clásica.")
    else:
        print("Por ahora no podemos ofrecerte una tarjeta.")

    continuar = input("¿Quieres evaluar a otra persona? (si/no): ")
    print("Programa realiado por Johan Paulo Nl = 04")
```
